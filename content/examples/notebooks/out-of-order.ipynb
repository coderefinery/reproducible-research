# Can a regional policy notebook reproduce its saved result?

## Goal
Understand why a believable saved result does not prove that the current notebook can recreate it.

## Setup
Imagine preparing a policy briefing with **fictional** unemployment rates:

| Fictional region | Rate |
| --- | --- |
| River | 20% |
| Hill | 22% |
| Coast | 24% |

The illustrative target is **10%**. Subtracting it from each rate gives gaps of 10, 12 and 14 **percentage points**. Their average is 12.0 percentage points above target. This average gives each region equal weight; it is not a pooled national unemployment rate. These are invented rates and a teaching target, not real statistics or an official policy goal.

Python's standard library is sufficient. All inputs are included; no data download is needed.

### A. Define the fictional regional rates
The list stores the rates for River, Hill and Coast, in that order.

```python
unemployment_rates = [20, 22, 24]
```

### B. Calculate the mean gap above the target
Subtract the target from each regional rate, then add the gaps and divide by the number of regions.

```python
gaps = [rate - target_rate for rate in unemployment_rates]
mean_gap = sum(gaps) / len(gaps)
print(f"Mean gap above target: {mean_gap:.1f} percentage points")
```

### C. Define the illustrative target
The target is a percentage rate; the calculated gaps are in percentage points.

```python
target_rate = 10
```
